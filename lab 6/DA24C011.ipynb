```python
#Loading required libraries
import unicodedata
import shutil
import os

import pandas as pd
import numpy as np

from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, classification_report
from sklearn.decomposition import PCA
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis as LDA
from sklearn.discriminant_analysis import QuadraticDiscriminantAnalysis as QDA
from sklearn.feature_extraction.text import CountVectorizer 
from sklearn.decomposition import TruncatedSVD

```

```python
from datasets import load_dataset
```

```python
#calling train data
dataset = load_dataset("qanastek/MASSIVE", split='train', trust_remote_code=True)
train_df=pd.DataFrame(dataset)
```

```python
#calling test data
test_df = load_dataset("qanastek/MASSIVE", split='test', trust_remote_code=True)
test_df = pd.DataFrame(test_df)
```

```python
#calling val data
val_df = load_dataset("qanastek/MASSIVE", split='validation', trust_remote_code=True)
val_df = pd.DataFrame(val_df)
```

```python
locales = ['af-ZA', 'da-DK', 'de-DE', 'en-US', 'es-ES', 'fr-FR', 'fi-FI', 'hu-HU','is-IS', 'it-IT', 'jv-ID',
           'lv-LV', 'ms-MY', 'nb-NO', 'nl-NL', 'pl-PL', 'pt-PT', 'ro-RO', 'ru-RU', 'sl-SL', 'sv-SE', 'sq-AL',
           'sw-KE', 'tl-PH', 'tr-TR', 'vi-VN','cy-GB']
```

```python
#creating files for train data
def deaccent(text):
    return ''.join(c for c in unicodedata.normalize('NFD', text) if unicodedata.category(c) != 'Mn')

for locale in locales:
    print("File currently loaded", locale)
    lang_data = dataset.filter(lambda example: example['locale'] == locale)

    utterances = [deaccent(utt) for utt in lang_data['utt']]
        
    with open(f"{locale}.txt", "w", encoding="utf-16") as f:
        for utt in utterances:
            f.write(utt + "\n")
print('filtered.')

local_save_path = os.path.join(os.getcwd(), 'MASSIVE_utterances_train')  # Change to your desired path

if not os.path.exists(local_save_path):
    os.makedirs(local_save_path)

for filename in os.listdir():
    if filename.endswith(".txt"):
        shutil.copy(filename, local_save_path)

print("Files saved to local directory successfully!")   
```

```python
#creating file for test data
for locale in locales:
    lang_data = test_df.filter(lambda example: example['locale'] == locale)

    utterances = [deaccent(utt) for utt in lang_data['utt']]
        
    with open(f"{locale}.txt", "w", encoding="utf-16") as f:
        for utt in utterances:
            f.write(utt + "\n")
print('filtered.')

local_save_path = os.path.join(os.getcwd(), 'MASSIVE_utterances_test')  

if not os.path.exists(local_save_path):
    os.makedirs(local_save_path)

for filename in os.listdir():
    if filename.endswith(".txt"):
        shutil.copy(filename, local_save_path)

print("Files saved to local directory successfully!")   
```

```python
#creating files for validation data
for locale in locales:
    lang_data = val_df.filter(lambda example: example['locale'] == locale)

    utterances = [deaccent(utt) for utt in lang_data['utt']]
        
    with open(f"{locale}.txt", "w", encoding="utf-16") as f:
        for utt in utterances:
            f.write(utt + "\n")
print('filtered.')

local_save_path = os.path.join(os.getcwd(), 'MASSIVE_utterances_validation')  

if not os.path.exists(local_save_path):
    os.makedirs(local_save_path)

for filename in os.listdir():
    if filename.endswith(".txt"):
        shutil.copy(filename, local_save_path)

print("Files saved to local directory successfully!")   
```

```python
#TASK 2: 
train_df = train_df[train_df['locale'].isin(locales)]
val_df = val_df[val_df['locale'].isin(locales)]
test_df = test_df[test_df['locale'].isin(locales)]
```

```python
#model building and training
vectorizer = CountVectorizer()

X_train = vectorizer.fit_transform(train_df['utt'])
y_train = train_df['locale']

X_val = vectorizer.transform(val_df['utt'])
y_val = val_df['locale']

X_test = vectorizer.transform(test_df['utt'])
y_test = test_df['locale']

nb_classifier = MultinomialNB()
nb_classifier.fit(X_train, y_train)
```

```python
#Evaluation on train data
y_train_pred = nb_classifier.predict(X_train)
train_accuracy = accuracy_score(y_train, y_train_pred)
print("Train Set Performance:")
print(f"Train Accuracy: {train_accuracy}")
print(classification_report(y_train, y_train_pred, target_names=locales))
```

```python
# Evaluation on the validation set
y_val_pred = nb_classifier.predict(X_val)
val_accuracy = accuracy_score(y_val, y_val_pred)
print("Validation Set Performance:")
print(f"Validation Accuracy: {val_accuracy}")
print(classification_report(y_val, y_val_pred, target_names=locales))
```

```python
# Evaluation on the test set
y_test_pred = nb_classifier.predict(X_test)
test_accuracy = accuracy_score(y_test, y_test_pred)
print("Test Set Performance:")
print(f"Test Accuracy: {test_accuracy}")
print(classification_report(y_test, y_test_pred, target_names=locales))
```

```python
#Creating RDA
class RegularizedDiscriminantAnalysis:
    def __init__(self, lambda_param=0.5):
        self.lambda_param = lambda_param
        self.lda = LDA()
        self.qda = QDA()

    def fit(self, X, y):
        self.lda.fit(X, y)
        self.qda.fit(X, y)

    def predict(self, X):
        lda_pred_prob = self.lda.predict_proba(X)
        qda_pred_prob = self.qda.predict_proba(X)
        combined_pred_prob = (1 - self.lambda_param) * lda_pred_prob + self.lambda_param * qda_pred_prob
        return np.argmax(combined_pred_prob, axis=1)


vectorizer = CountVectorizer(min_df=5)
X_train = vectorizer.fit_transform(train_df['utt'])
X_test = vectorizer.transform(test_df['utt'])  

# Dimensionality reduction 
n_components = 100
svd = TruncatedSVD(n_components=n_components)
X_train_reduced = svd.fit_transform(X_train)
X_test_reduced = svd.transform(X_test)


y_train = train_df['locale']
y_test = test_df['locale']

acc = 0
lambdas = [0.1,0.2,0.25,0.5]
for l in lambdas:
    rda_model = RegularizedDiscriminantAnalysis(l)
    rda_model.fit(X_train_reduced, y_train)
    y_pred = rda_model.predict(X_test_reduced)
    accuracy = accuracy_score(y_test, y_pred)
    if accuracy > acc:
        acc = accuracy
        best_lambda = l


print(f"Best lambda: {l}, accuracy: {acc:0.2f}")
```
